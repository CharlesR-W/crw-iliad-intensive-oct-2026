## 4.1 Stagewise development and phase transitions

*What this is:* Notebook 2 §2.3 said the posterior changes basin at critical sample
sizes. These papers track $\hat\lambda$ along a training run and use its
plateaus and jumps to divide training into stages.

<div class="reading">

**[Dynamical versus Bayesian Phase Transitions in a Toy Model of Superposition](https://arxiv.org/abs/2310.06301)**, Chen, Lau, Mendel, Wei & Murfet (2023) <span class="ratings"><span class="rate rate-imp" title="Importance 4/5: recommended: one of the best readings for this section">Importance <span class="on">●●●●</span><span class="off">○</span></span> <span class="rate rate-len" title="Length 3/5: 1-2 hours (a typical paper, read through)">Length <span class="on">●●●</span><span class="off">○○</span></span></span>

In a small autoencoder whose critical points are regular polygons, the learning coefficients of those polygons predict the posterior's phase transitions in $n$, and the same critical points organize what SGD training does.

</div>

<div class="reading">

**[Loss Landscape Degeneracy and Stagewise Development in Transformers](https://arxiv.org/abs/2402.02364)**, Hoogland, Wang, Farrugia-Roberts, Carroll, Wei & Murfet (TMLR 2025) <span class="ratings"><span class="rate rate-imp" title="Importance 4/5: recommended: one of the best readings for this section">Importance <span class="on">●●●●</span><span class="off">○</span></span> <span class="rate rate-len" title="Length 3/5: 1-2 hours (a typical paper, read through)">Length <span class="on">●●●</span><span class="off">○○</span></span></span>

## 4.2 Susceptibilities

*What this is:* in Notebook 1's dictionary, a derivative of a posterior expectation
with respect to a small change of the "Hamiltonian" is a susceptibility, and
it equals a posterior covariance. Here the change is a reweighting of the
data distribution.

<div class="reading">

**[Structural Inference: Interpreting Small Language Models with Susceptibilities](https://arxiv.org/abs/2504.18274)**, Baker, Wang, Hoogland & Murfet (2025) <span class="ratings"><span class="rate rate-imp" title="Importance 3/5: worthwhile: for most people taking this section">Importance <span class="on">●●●</span><span class="off">○○</span></span> <span class="rate rate-len" title="Length 3/5: 1-2 hours (a typical paper, read through)">Length <span class="on">●●●</span><span class="off">○○</span></span></span>

Shifting the data distribution slightly and measuring the first-order response of a local posterior expectation, estimated with SGLD samples, gives per-component and per-token attribution scores whose low-rank structure separates functional modules of a small transformer.

</div>

<div class="reading">

**[Towards Spectroscopy: Susceptibility Clusters in Language Models](https://arxiv.org/abs/2601.12703)**, Gordon, Baker, Wang, Snell, van Wingerden & Murfet (2026) <span class="ratings"><span class="rate rate-imp" title="Importance 3/5: worthwhile: for most people taking this section">Importance <span class="on">●●●</span><span class="off">○○</span></span> <span class="rate rate-len" title="Length 3/5: 1-2 hours (a typical paper, read through)">Length <span class="on">●●●</span><span class="off">○○</span></span></span>

Susceptibilities to upweighting a token in a context, computed as covariances under the localized tempered posterior, cluster tokens that are predicted for similar reasons; applied to a 14M-parameter model.

</div>

## 4.3 Bayesian influence functions

*What this is:* classical influence functions ask how a trained parameter
moves when one training point is reweighted, and need an inverse Hessian;
the Bayesian version asks how a posterior expectation moves, which is again
a covariance under the local posterior.

<div class="reading">

**[Bayesian Influence Functions for Hessian-Free Data Attribution](https://arxiv.org/abs/2509.26544)**, Kreer, Wu, Adam, Furman & Hoogland (2025) <span class="ratings"><span class="rate rate-imp" title="Importance 3/5: worthwhile: for most people taking this section">Importance <span class="on">●●●</span><span class="off">○○</span></span> <span class="rate rate-len" title="Length 3/5: 1-2 hours (a typical paper, read through)">Length <span class="on">●●●</span><span class="off">○○</span></span></span>

Replaces the inverse-Hessian step of the classical influence function by a covariance of per-sample losses under the local tempered posterior, estimated by SGLD, so that it is defined when the Hessian is singular and scales to large models.

</div>

<div class="reading">

**[The Loss Kernel: A Geometric Probe for Deep Learning Interpretability](https://arxiv.org/abs/2509.26537)**, Adam, Furman & Hoogland (2025) <span class="ratings"><span class="rate rate-imp" title="Importance 2/5: optional: more on one specific point">Importance <span class="on">●●</span><span class="off">○○○</span></span> <span class="rate rate-len" title="Length 3/5: 1-2 hours (a typical paper, read through)">Length <span class="on">●●●</span><span class="off">○○</span></span></span>

The covariance matrix of per-sample losses under low-loss-preserving perturbations of the parameters, used as a similarity between data points and compared with semantic hierarchies on ImageNet.

</div>

## 4.4 Minimum description length and compressibility

*What this is:* Notebook 1 §1.3 read the free energy as a code length; $\lambda$ is
then the coefficient of $\log n$ in the description length of the model, and
the question is whether it predicts how well a trained network compresses.

<div class="reading">

**[Compressibility Measures Complexity: Minimum Description Length Meets Singular Learning Theory](https://arxiv.org/abs/2510.12077)**, Urdshals, Lau, Hoogland, van Wingerden & Murfet (2025) <span class="ratings"><span class="rate rate-imp" title="Importance 3/5: worthwhile: for most people taking this section">Importance <span class="on">●●●</span><span class="off">○○</span></span> <span class="rate rate-len" title="Length 3/5: 1-2 hours (a typical paper, read through)">Length <span class="on">●●●</span><span class="off">○○</span></span></span>

Extends the minimum description length principle to models with degenerate minima and reports, across the Pythia models and several compression schemes, that $\hat\lambda$ tracks (sometimes linearly) how compressible a network is.

</div>
